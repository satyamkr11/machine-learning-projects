# Machine Learning Practical — Logistic Regression

**Objective:** Predict Gender (F/M) from the preference attributes in `Transformed Data Set - Sheet1.csv` using Logistic Regression.

This notebook follows the teacher-provided workflow: data loading → preprocessing → train/test split → scaling → Logistic Regression → evaluation.

## 1. Load the dataset

If the CSV is already uploaded to Colab, it will be used automatically. Otherwise, Colab will open an upload dialog.

```python
import numpy as np
import pandas as pd
import os

file_name = 'Transformed Data Set - Sheet1.csv'
file_path = f'/content/{file_name}'

if not os.path.exists(file_path):
    from google.colab import files
    uploaded = files.upload()
    # Find the uploaded CSV even if Colab renamed it slightly
    csv_files = [name for name in uploaded.keys() if name.lower().endswith('.csv')]
    if not csv_files:
        raise FileNotFoundError('Please upload Transformed Data Set - Sheet1.csv')
    file_path = f'/content/{csv_files[0]}'

df = pd.read_csv(file_path)
print(f'Dataset loaded: {file_path}')
df.head()
```

## 2. Convert Gender to numerical values

`F → 0` and `M → 1`, as required before training the model.

```python
df['Gender'] = np.where(df['Gender'] == 'F', 0, 1)
df.head()
```

```python
df['Gender'].value_counts()
```

## 3. Encode categorical features

The preference columns are categorical text. One-hot encoding converts them into numerical indicator columns. `drop_first=True` removes the first category from each encoded group.

```python
df = pd.get_dummies(df, drop_first=True)
df.head()
```

## 4. Separate features (X) and target (y)

```python
x = df.drop('Gender', axis=1)
y = df['Gender']

print('X shape:', x.shape)
print('y shape:', y.shape)
x.head()
```

```python
y.head()
y.value_counts()
```

## 5. Split into training and testing data

80% of the data is used for training and 20% for testing. `stratify=y` keeps the class proportions balanced.

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42, stratify=y
)

print('Training samples:', len(x_train))
print('Testing samples:', len(x_test))
```

## 6. Feature scaling

StandardScaler puts the input features on a comparable scale. The scaler is fitted only on the training data and then applied to the test data.

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

## 7. Train the Logistic Regression model

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(x_train, y_train)
```

## 8. Evaluate the model

We calculate the confusion matrix, accuracy, and classification report.

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report

y_pred = model.predict(x_test)

cm = confusion_matrix(y_test, y_pred)
accuracy = accuracy_score(y_test, y_pred)

print('Confusion Matrix:')
print(cm)
print(f'\nAccuracy: {accuracy:.4f} ({accuracy*100:.2f}%)')
print('\nClassification Report:')
print(classification_report(y_test, y_pred, target_names=['Female (0)', 'Male (1)']))
```
