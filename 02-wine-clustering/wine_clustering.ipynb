# Machine Learning Practical — Clustering

**Objective:** Perform clustering on `wine-clustering.csv` using hierarchical clustering and inspect the number of clusters with a dendrogram and K-Means elbow curve.

This notebook follows the teacher-provided workflow and uses the 13 numerical wine features.

## 1. Load the dataset

If the CSV is already uploaded to Colab, it will be used automatically. Otherwise, Colab will open an upload dialog.

```python
import numpy as np
import pandas as pd
import os

file_name = 'wine-clustering.csv'
file_path = f'/content/{file_name}'

if not os.path.exists(file_path):
    from google.colab import files
    uploaded = files.upload()
    csv_files = [name for name in uploaded.keys() if name.lower().endswith('.csv')]
    if not csv_files:
        raise FileNotFoundError('Please upload wine-clustering.csv')
    file_path = f'/content/{csv_files[0]}'

df = pd.read_csv(file_path)
print(f'Dataset loaded: {file_path}')
print('Shape:', df.shape)
df.head()
```

## 2. Prepare the feature matrix

All columns in this dataset are numerical wine measurements, so they are used as clustering features.

```python
x = df.iloc[:, :].values
print('Feature matrix shape:', x.shape)
```

## 3. Hierarchical clustering — Dendrogram

Ward linkage groups observations by minimizing within-cluster variance. The dendrogram helps visualize a reasonable number of clusters.

```python
import matplotlib.pyplot as plt
import scipy.cluster.hierarchy as sch

plt.figure(figsize=(12, 6))
dendrogram = sch.dendrogram(sch.linkage(x, method='ward'))
plt.title('Dendrogram')
plt.xlabel('Wine Samples')
plt.ylabel('Euclidean Distance')
plt.show()
```

## 4. K-Means elbow method

We calculate inertia for k = 1 to 10. The elbow in the curve is a common guide for choosing a cluster count.

```python
from sklearn.cluster import KMeans

k_range = range(1, 11)
inertia_values = []

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(x)
    inertia_values.append(kmeans.inertia_)

plt.figure(figsize=(8, 4))
plt.plot(list(k_range), inertia_values, marker='o')
plt.title('Elbow Method for K-Means')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.xticks(list(k_range))
plt.show()
```

## 5. Agglomerative Hierarchical Clustering

The teacher notebook uses **3 clusters** with Ward linkage.

```python
from sklearn.cluster import AgglomerativeClustering

hc = AgglomerativeClustering(n_clusters=3, linkage='ward')
y_hc = hc.fit_predict(x)

print('Cluster counts:')
print(pd.Series(y_hc).value_counts().sort_index())
```

```python
df['hc_y'] = y_hc
df.head()
```

## 6. Visualise the clusters

The teacher notebook plots the first two features and colours the three hierarchical-clustering groups.

```python
plt.figure(figsize=(8, 6))
plt.scatter(x[y_hc == 0, 0], x[y_hc == 0, 1], s=100, label='Cluster 0')
plt.scatter(x[y_hc == 1, 0], x[y_hc == 1, 1], s=100, label='Cluster 1')
plt.scatter(x[y_hc == 2, 0], x[y_hc == 2, 1], s=100, label='Cluster 2')
plt.xlabel(df.columns[0])
plt.ylabel(df.columns[1])
plt.title('Hierarchical Clustering — First Two Features')
plt.legend()
plt.show()
```
